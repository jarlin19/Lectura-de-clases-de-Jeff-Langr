```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# URL del dataset
url = "https://raw.githubusercontent.com/hmedrano1979/ITC/refs/heads/main/netflix_titles.csv"

# Cargar el dataset
df = pd.read_csv(url)

# Mostrar las primeras filas
df.head()
```

El dataset "netflix_titles.csv" contiene información sobre el catálogo de Netflix. Cada fila representa una película o serie y sus atributos como:

ID: Identificador único.
Tipo: Especifica si es una "Movie" o una "TV Show".
Título: Nombre de la película o serie.
Director: Nombre del director (si aplica).
Elenco: Lista de actores principales.
País: País de origen de la producción.
Fecha de Adición: Cuándo se agregó al catálogo de Netflix.
Año de lanzamiento: Año en que se estrenó.
Clasificación: Tipo de audiencia recomendada.
Duración: Duración en minutos (películas) o número de temporadas (series).
Géneros: Categorías en las que está clasificado el contenido.
Descripción: Breve sinopsis.
Relevancia del dataset: Este dataset es útil para analizar tendencias de contenido en Netflix, distribución geográfica, géneros más populares, y más.

 Desarrollo del Análisis Exploratorio
Películas vs. Series

```python
sns.countplot(data=df, x='type')
plt.title("Cantidad de Películas vs. Series en Netflix")
plt.show()
```

Distribución de Clasificaciones

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, y="rating", order=df['rating'].value_counts().index)
plt.title("Distribución de Clasificaciones en Netflix")
plt.show()
```

Evolución de Contenidos en Netflix

```python
plt.figure(figsize=(12,6))
df['release_year'].value_counts().sort_index().plot(kind='line')
plt.title("Cantidad de Producciones por Año")
plt.xlabel("Año")
plt.ylabel("Cantidad")
plt.show()
```

Conclusiones
Netflix tiene más películas que series, pero las series han crecido en número en los últimos años.
Las clasificaciones más comunes son 'TV-MA' y 'PG-13', lo que indica que hay mucho contenido para adultos jóvenes.
El contenido en Netflix ha crecido significativamente en los últimos 10 años, mostrando una tendencia a producir más contenido original.
